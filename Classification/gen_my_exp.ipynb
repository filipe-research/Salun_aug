```python
forget = {
    'cifar100': {10: 4500, 50: 22500},
    'cifar10': {10: 4500, 50: 22500}
}

augs = {
    'default': "crop-flip",
    'reraser': "crop-flip-rerase",
    'randaug' : 'crop-flip-randaug',
    'autoaug' : "crop-flip-autoaug",
    'triviaug': "crop-flip-trivial",
    'augmix': "crop-flip-augmix",
    'noaug': None
}
```

```python
#  parameters
#cifar10/ cifar100
dataset = "cifar100" 
save_root = "myexps"
# default/ reraser/ randaug/ autoaug/ triviaug/ augmix/ noaug
augmentation = "reraser"

# baseline/ retrain/  salun/ rl / ft
method = "rl"
forget_rate = 10

# dataset_path = "/mnt/hd_pesquisa/pesquisa/datasets/"
dataset_path = "/home/pesquisador/pesquisa/datasets"

```

```python
#random forget code

for i in range(1,6):
    save_dir = f"{save_root}/exp_{dataset}_{method}_df{forget_rate}_{augmentation}_{i}"
    baseline_dir = f"exp_{dataset}_baseline_{augmentation}_{i}"
    model_path = f"{save_root}/{baseline_dir}/0model_SA_best.pth.tar"
    seed = 10*i

       
    if method == "salun":

        if augmentation == "noaug" :
            command1 = f"python3 generate_mask.py --dataset {dataset} --save_dir {save_dir} --model_path {model_path} --unlearn_epochs 1 --num_indexes_to_replace {forget[dataset][forget_rate]} --data {dataset_path} --no-aug --seed {seed} --train_seed {seed};"
            command2 = f"python3 main_random.py --unlearn RL --unlearn_epochs 10 --unlearn_lr 0.013 --model_path {model_path} --save_dir {save_dir} --mask_path {save_dir}/with_0.5.pt --num_indexes_to_replace {forget[dataset][forget_rate]} --dataset {dataset} --data {dataset_path}  --no-aug --seed {seed} --train_seed {seed};"
        else:
            command1 = f"python3 generate_mask.py --dataset {dataset} --save_dir {save_dir} --model_path {model_path} --unlearn_epochs 1 --num_indexes_to_replace {forget[dataset][forget_rate]} --data {dataset_path} --aug_mode {augs[augmentation]} --seed {seed} --train_seed {seed};"
            command2 = f"python3 main_random.py --unlearn RL --unlearn_epochs 10 --unlearn_lr 0.013 --model_path {model_path} --save_dir {save_dir} --mask_path {save_dir}/with_0.5.pt --num_indexes_to_replace {forget[dataset][forget_rate]} --dataset {dataset} --data {dataset_path}  --aug_mode {augs[augmentation]} --seed {seed} --train_seed {seed};"
        print(command1)
        print(command2)
        print("\n")
    elif method == "rl":

        if augmentation == "noaug" :
            command = f"python3 main_forget.py --unlearn RL --unlearn_epochs 10 --unlearn_lr 0.013 --model_path {model_path} --save_dir {save_dir}  --num_indexes_to_replace {forget[dataset][forget_rate]} --dataset {dataset} --data {dataset_path}  --no-aug --seed {seed} --train_seed {seed};"
        else:
            command = f"python3 main_forget.py --unlearn RL --unlearn_epochs 10 --unlearn_lr 0.013 --model_path {model_path} --save_dir {save_dir}  --num_indexes_to_replace {forget[dataset][forget_rate]} --dataset {dataset} --data {dataset_path}  --aug_mode {augs[augmentation]} --seed {seed} --train_seed {seed};"
        print(command)
        print("\n")

    elif method == "ft":

        if augmentation == "noaug" :
            command = f"python3 main_forget.py --unlearn FT --unlearn_epochs 10 --unlearn_lr 0.013 --model_path {model_path} --save_dir {save_dir}  --num_indexes_to_replace {forget[dataset][forget_rate]} --dataset {dataset} --data {dataset_path}  --no-aug --seed {seed} --train_seed {seed};"
        else:
            command = f"python3 main_forget.py --unlearn FT --unlearn_epochs 10 --unlearn_lr 0.013 --model_path {model_path} --save_dir {save_dir}  --num_indexes_to_replace {forget[dataset][forget_rate]} --dataset {dataset} --data {dataset_path}  --aug_mode {augs[augmentation]} --seed {seed} --train_seed {seed};"
        print(command)
        print("\n")


    elif method == "baseline":
        save_dir = f"{save_root}/exp_{dataset}_{method}_{augmentation}_{i}"

        if augmentation == "noaug" :
            command = f"python main_train.py --arch resnet18 --dataset {dataset} --lr 0.1 --epochs 200 --data {dataset_path} --save_dir {save_dir}  --no-aug --seed {seed} --train_seed {seed};"
        else:
         
            command = f"python main_train.py --arch resnet18 --dataset {dataset} --lr 0.1 --epochs 200 --data {dataset_path} --save_dir {save_dir} --aug_mode {augs[augmentation]} --seed {seed} --train_seed {seed};"
        print(command)
        print("\n")
    elif method == "retrain":
        if augmentation == "noaug" :
            command = f"python3 main_forget.py --save_dir {save_dir} --model_path {model_path} --unlearn retrain --num_indexes_to_replace {forget[dataset][forget_rate]} --unlearn_epochs 200 --unlearn_lr 0.1    --dataset {dataset} --data {dataset_path}  --no-aug --seed {seed} --train_seed {seed};"
        else:
            command = f"python3 main_forget.py --save_dir {save_dir} --model_path {model_path} --unlearn retrain --num_indexes_to_replace {forget[dataset][forget_rate]} --unlearn_epochs 200 --unlearn_lr 0.1    --dataset {dataset} --data {dataset_path}  --aug_mode {augs[augmentation]} --seed {seed} --train_seed {seed};"
        print(command)
        print("\n")
    else:
        print("invalido")

```
